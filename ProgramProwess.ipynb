# Main Program

```python
# These are imported libraries that the program needs to run.
from sqlalchemy import create_engine
from IPython.display import display
import pandas as pd
import sqlite3
import datetime as dt

# These are the variables that'll contain the necessary components of the main program.
data = pd.read_csv('ProgramProwess Datasheet.csv')
df = pd.DataFrame(data, columns=['Date', 'Time Spent', 'Language(s)', 'Improvements', 'Struggles'])
conn = sqlite3.connect('programmingdata.db')
cursor = conn.cursor()
engine = create_engine('sqlite:///programmingdata.db', echo=False)
df.to_sql('Programming_Data', con=engine, if_exists='append')

# This is the area in which the main program will be assembled.
def mainMenu():
  print('Select one of the following choices:')
  while True:
    print ("1. Log Entry")
    print ('2. Delete Entry')
    print ('3. Entry Index')
    print ('4. Summaries')
    print ('5. Exit')
    try:
      selection = int(input())
      if selection == 1:
        print('------------')
        print('What is the date? Preferrably enter in the following format: MM/DD/YYYY')
        curDate = input()
        print('How long did you spend programming today? Preferrably enter in the following format: h:m:s')
        curTime = input()
        print('What language(s) did you use today?')
        curLanguages = input()
        print('Any improvements?')
        curImprovements = input()
        print('Any struggles?')
        curStruggles = input()
        if curDate == '' and curTime == '' and curLanguages == '' and curImprovements == '' and curStruggles == '':
          print('You cannot enter an entry with no values.')
        else:
          df.loc[len(df)] = [curDate, curTime, curLanguages, curImprovements, curStruggles]
          df.to_csv('ProgramProwess Datasheet.csv', mode='w+')
          df.to_sql('Programming_Data', con=engine, if_exists='append')
          print('Entry added!')
        print('------------')
      elif selection == 2:
        print('------------')
        try:
          print ('Which entry would you like to remove? Please provide the NUMERICAL row (i.e. 0 for the first entry).')
          markedDelete = int(input())
          df.drop(markedDelete, inplace=True)
          df.to_csv('ProgramProwess Datasheet.csv', mode='w+')
          df.to_sql('Programming_Data', con=engine, if_exists='append')
          print('Entry deleted!')
          print('------------')
        except:
          print('------------')
          print('You entered in an invalid row. Try again.')
          print('------------')
      elif selection == 3:
        print('------------')
        display(df)
        print('------------')
      elif selection == 4:
        print('------------')
        try:
          if df.loc[len(df) - 1, 'Improvements'] == "":
            print('You have no improvements as of late.')
          else:
            print('Recently, you seem to be doing well with ' + df.loc[len(df) - 1, 'Improvements'] + '.')
        except:
          print('You have no improvements as of late.')
        try:
          if df.loc[len(df) - 1, 'Struggles'] == "":
            print('You have no struggles as of late!')
          else:
            print('You seem to be struggling with ' + df.loc[len(df) - 1, 'Struggles'] + ', however.')
        except:
          print('You have no struggles as of late!')
        print('------------')
      elif selection == 5:
        print('------------')
        print('Goodbye!')
        print('------------')
        break
      else:
        print('Not a valid option. Try again.')
        print('------------')
    except:
      print('Something went wrong. Try again.')
      print('------------')

# This is essentially the program assembled in 'full'.
print ('Welcome to ProgramProwess!')
print('------------')
print(f'Today is {dt.date.today().strftime("%B %d, %Y")}...')
print('------------')
mainMenu()
```

# Program Is Finished When...

1. User can input a value for each respective column in the datasheet
2. User CANNOT input a non-valid input for any column in the datasheet (i.e null for any column)
3. User CANNOT input a non-valid input for a menu option (i.e. any number other than 1 through 4, null, etc.)
4. User CANNOT insert an entry with NO values (one value filled in and the rest blank is fine)
5. The summary DOES NOT factor in empty information from the datasheet

1. Ability to insert daily programming progress into a designated datasheet
2. Ability to view each entry saved up to that day
3. Ability to review improvements and struggles most recent to the user as a summary
